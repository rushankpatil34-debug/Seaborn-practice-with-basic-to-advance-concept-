```python
import pandas as pd
# 1. Define the data dictionary
data = {
    "Deprement": ["AI", "CS", "CS", "ML", "AI", "CS", "IT"],
    "Gender": ["male", "female", "male", "female", "male", "female", "female"],
}

# 2. Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# 3. Create the count plot using the correct DataFrame variable (df)
sns.countplot(data=df, x="Deprement", hue="Gender")

# 4. Add a title and display the plot
plt.title("Number of Students by Deprement")
plt.show()
```

```python
data={
"Departments": ["IT","IT", "IT", "CS", "CS", "CS"],

"Marks": [70,80,90,80,90,100]

}

df = pd.DataFrame(data)


sns.barplot(data = df, x = "Departments", y = "Marks", estimator="median")

plt.title("Average Marks by Deparment")

plt.show()
```

```python
data = {
    "Category": [
        "Technology",
        "Furniture",
        "Office Supplies",
        "Technology",
        "Furniture",
        "Office Supplies",
        "Technology",
        "Furniture"
    ],
    "Sales": [
        1200,
        800,
        500,
        1500,
        900,
        600,
        1800,
        700
    ],
    "Profit": [
        300,
        100,
        80,
        450,
        150,
        120,
        500,
        90
    ],
    "Regions":[
        "south",
        "North",
        "West",
        "East",
        "south",
        "North",
        "West",
        "East"
    ]
}

df = pd.DataFrame(data)
print(df)
```

```python
sns.barplot(data = df, x = "Category", y = "Sales", estimator = "sum")
plt.title("Average Sales by category")

plt.show()
```

```python
sns.countplot(data = df, x = "Category")
plt.title("Number of products by category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
sns.barplot(data = df, x = "Category", y = "Sales")
plt.title("median Sales by category")

plt.show()
```

```python
sns.boxplot(data = df, x = "Category", y= "Sales")
plt.title("Seals Distribution by category")

plt.show()
```

```python
sns.violinplot(data = df, x = "Category", y= "Sales")
plt.show()
```

```python
sns.stripplot(data = df, x = "Category", y= "Sales", hue = "Regions")
plt.show()
```

```python
sns.swarmplot(data = df, x = "Category", y= "Sales", hue = "Regions")
plt.show()
```

```python
sns.catplot(data = df, x = "Category", y= "Sales", hue = "Regions", kind="bar")
#{kind}=[strip,sworm,box,violin,point,bar,count.]
plt.show()
```
